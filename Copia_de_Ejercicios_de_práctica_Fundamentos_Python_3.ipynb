**18.Función que recibe una lista**

```python
def promedio(notas):
    total = 0
    for n in notas:
        total = total + n
    return total /len(notas)
print(promedio([3.5, 4.2, 2.8]))
```

**Reto integrador**

**20.Reporte de notas con funciones**

```python
def promedio(notas):
    total = 0
    for n in notas:
        total = total + n
    return total / len(notas)
def reporte(nombre, notas):
    prom = promedio(notas)
    print(nombre, round(prom, 2))
    if prom >= 3.0:
        print("Aprobado")
    else:
        print("No aprobado")
reporte("Laura", [3.5, 4.2, 2.8])
```
