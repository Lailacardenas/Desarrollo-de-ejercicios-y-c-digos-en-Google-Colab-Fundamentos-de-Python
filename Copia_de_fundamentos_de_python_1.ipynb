**Clasificación de temperaturas**

```python
# El programa debe solicitar 5 temperaturas.
# Para cada una debe mostrar:
#   "Baja"   si t < 18
#   "Normal" si 18 <= t <= 25
#   "Alta"   si t > 25
for i in range(5):
    t = float(input("Temperatura: "))
    if t < 18:
        print("Baja")
    elif t <= 25:
        print("Normal")
    else:
        print("alta")
```
